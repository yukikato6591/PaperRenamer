```python
from pathlib import Path
import pdf_renamer

folder = Path("Path")

for pdf in folder.rglob("*.pdf"):
    info = pdf_renamer.extract_from_pdf(pdf)
    new_name = pdf_renamer.build_filename(info)

    if new_name == "untitled.pdf":
        print(f"SKIP: {pdf.name}")
        continue

    dst = pdf_renamer.unique_path(pdf.with_name(new_name))
    print(f"{pdf.name} -> {dst.name}")
    pdf.rename(dst)
```
